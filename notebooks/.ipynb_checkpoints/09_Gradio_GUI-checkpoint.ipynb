## What is Gradio?

**Gradio** is a Python library that creates web interfaces with just a few lines of code. No HTML, CSS, or JavaScript needed!

With Gradio, we can create:
- File upload interfaces
- Chat interfaces
- Form-based interfaces
- All with pure Python

## Why Gradio for This Project?

- No frontend development needed
- Works locally without a web server
- Perfect for demos and prototypes
- Easy to share with others

## Step 1: Import Required Libraries

```
import gradio as gr
from pydantic import BaseModel, Field
from typing import Literal
from langchain_ollama import OllamaLLM
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import PydanticOutputParser
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma
from langchain_core.runnables import RunnablePassthrough
import os
import shutil

print("All imports successful!")
```

## Step 2: Create the RAG Pipeline Class

We'll wrap everything into a class for clean code organization.

```
class PDFChatAssistant:
    """
    A simple RAG-based PDF chat assistant.
    Loads a PDF, creates embeddings, and answers questions.
    """
    
    def __init__(self):
        """Initialize the assistant (does not load PDF yet)."""
        self.chunks = None
        self.vectorstore = None
        self.retriever = None
        self.llm = None
        self.chain = None
        self.current_pdf_path = None
        
    def load_pdf(self, pdf_path):
        """Load a PDF file and create the RAG pipeline."""
        print(f"Loading PDF: {pdf_path}")
        
        # Load the PDF
        loader = PyPDFLoader(pdf_path)
        pages = loader.load()
        
        # Split into chunks
        text_splitter = RecursiveCharacterTextSplitter(
            chunk_size=500, chunk_overlap=100
        )
        self.chunks = text_splitter.split_documents(pages)
        
        # Create embeddings
        embeddings = HuggingFaceEmbeddings(
            model_name="sentence-transformers/all-MiniLM-L6-v2",
            model_kwargs={"device": "cpu"},
            encode_kwargs={"normalize_embeddings": True}
        )
        
        # Clear old vector store
        db_path = "../chroma_db"
        if os.path.exists(db_path):
            shutil.rmtree(db_path)
        
        # Create vector store
        self.vectorstore = Chroma.from_documents(
            documents=self.chunks,
            embedding=embeddings,
            persist_directory=db_path
        )
        
        # Create retriever
        self.retriever = self.vectorstore.as_retriever(search_kwargs={"k": 4})
        
        # Create LLM
        self.llm = OllamaLLM(
            model="mistral", temperature=0.1, num_ctx=4096
        )
        
        # Create output parser
        class RAGResponse(BaseModel):
            answer: str = Field(description="The answer to the question")
            source_page: str = Field(description="Page number")
            confidence: Literal["high", "medium", "low", "not_found"] = Field(
                description="Confidence level"
            )
            retrieved_context: str = Field(description="Context used")
        
        self.parser = PydanticOutputParser(pydantic_object=RAGResponse)
        format_instructions = self.parser.get_format_instructions()
        
        # Create prompt
        self.prompt = PromptTemplate.from_template(
            "You are a document analyst. Answer using ONLY the provided context.\n\n"
            "## Context:\n{context}\n\n"
            "## Question:\n{question}\n\n"
            "## Rules:\n"
            "1. Use ONLY the provided context.\n"
            "2. If the answer is not in the context, set confidence to 'not_found'.\n"
            "3. Do not hallucinate.\n\n"
            "{format_instructions}"
        ).partial(format_instructions=format_instructions)
        
        # Build chain
        self.chain = (
            {"context": self.retriever | self._format_docs, "question": RunnablePassthrough()}
            | self.prompt
            | self.llm
            | self.parser
        )
        
        self.current_pdf_path = pdf_path
        return f"PDF loaded: {len(pages)} pages, {len(self.chunks)} chunks"
    
    def _format_docs(self, docs):
        """Format retrieved documents into a context string."""
        return "\n\n".join([
            f"[{i+1}] (Page {doc.metadata.get('page', '?')})\n{doc.page_content}"
            for i, doc in enumerate(docs)
        ])
    
    def ask_question(self, question):
        """Ask a question and return the structured answer."""
        if not self.chain:
            return "Please upload a PDF first."
        
        try:
            result = self.chain.invoke(question)
            return (
                f"**Answer:** {result.answer}\n\n"
                f"**Source Page:** {result.source_page}\n\n"
                f"**Confidence:** {result.confidence}\n\n"
                f"**Retrieved Context:**\n{result.retrieved_context}"
            )
        except Exception as e:
            # Fallback: use simpler chain without parser
            try:
                simple_prompt = PromptTemplate.from_template(
                    "Answer using ONLY the context below.\n\n"
                    "## Context:\n{context}\n\n"
                    "## Question:\n{question}\n\n"
                    "## Answer:\n"
                )
                simple_chain = (
                    {"context": self.retriever | self._format_docs, "question": RunnablePassthrough()}
                    | simple_prompt
                    | self.llm
                )
                answer = simple_chain.invoke(question)
                return f"**Answer:** {answer}\n\n**Source Page:** (See retrieved context below)\n\n**Retrieved Context:**\n{self._format_docs(self.retriever.invoke(question))}"
            except Exception as e2:
                return f"Error: {e2}"

# Test the class
assistant = PDFChatAssistant()
print("PDFChatAssistant class created!")
```

## Step 3: Create the Gradio Interface

We'll create a simple interface with:
- PDF file upload
- Question text box
- Ask button
- Output area showing answer, source page, and context

```
# Create the Gradio interface
def ask_question_fn(question, pdf_file):
    """Function that Gradio will call when user clicks Ask."""
    if pdf_file is None:
        return "Please upload a PDF file first."
    
    if assistant.current_pdf_path != pdf_file:
        assistant.load_pdf(pdf_file)
    
    if not question.strip():
        return "Please enter a question."
    
    return assistant.ask_question(question)

# Build the Gradio interface
with gr.Blocks(title="AI PDF Chat Assistant") as demo:
    gr.Markdown("# AI PDF Chat Assistant")
    gr.Markdown("Upload a PDF and ask questions about its content.")
    
    with gr.Row():
        with gr.Column():
            # PDF upload
            pdf_file = gr.File(label="Upload PDF", file_types=[".pdf"])
            
            # Question input
            question = gr.Textbox(
                label="Your Question",
                placeholder="Ask a question about the PDF...",
                lines=2
            )
            
            # Ask button
            ask_btn = gr.Button("Ask", variant="primary")
        
        with gr.Column():
            # Output display
            output = gr.Textbox(label="Response", lines=10)
    
    # Connect the button to the function
    ask_btn.click(
        fn=ask_question_fn,
        inputs=[question, pdf_file],
        outputs=[output]
    )

print("Gradio interface created!")
```

## Key Takeaways

1. **Gradio** creates web interfaces with minimal Python code
2. **File upload** is handled automatically by Gradio
3. **gr.Blocks()** gives us layout control
4. **gr.Row()** and **gr.Column()** organize the layout
5. The **click** method connects buttons to functions

## GUI Components

| Component | Purpose |
|-----------|----------|
| gr.File | Upload PDF |
| gr.Textbox | Enter question |
| gr.Button | Trigger action |
| gr.Textbox (output) | Display results |
