## Project Overview

The **AI PDF Chat Assistant** is a local application that:

1. Accepts a PDF upload
2. Extracts text from the PDF
3. Splits the text into chunks
4. Creates embeddings using HuggingFace
5. Stores embeddings in ChromaDB
6. Accepts questions from the user
7. Retrieves relevant chunks from ChromaDB
8. Sends context + question to Mistral (via Ollama)
9. Returns a structured answer with source page and confidence

## Architecture Diagram

```
User uploads PDF
    |
    v
[1] PyPDFLoader -> Extract text
    |
    v
[2] RecursiveCharacterTextSplitter -> Split into chunks
    |
    v
[3] HuggingFaceEmbeddings -> Create vectors
    |
    v
[4] ChromaDB -> Store vectors
    |
    v
User asks question
    |
    v
[5] Embed question -> Search ChromaDB -> Get relevant chunks
    |
    v
[6] Build prompt (context + question + instructions)
    |
    v
[7] Ollama + Mistral -> Generate answer
    |
    v
[8] Pydantic Output Parser -> Structure response
    |
    v
[9] Gradio GUI -> Display to user
```

## Step 1: Import All Required Libraries

````
import os
import shutil
import gradio as gr
from pydantic import BaseModel, Field
from typing import Literal
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma
from langchain_ollama import OllamaLLM
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import PydanticOutputParser
from langchain_core.runnables import RunnablePassthrough

print("All imports successful!")
print(f"LangChain version: imported")
print(f"Project: AI PDF Chat Assistant")
````

## Step 2: Define the Output Schema

````
class RAGResponse(BaseModel):
    """
    Structured output schema for RAG responses.
    
    This defines the exact format we expect from the LLM.
    The PydanticOutputParser will convert the LLM's text output
    into this structured format.
    """
    answer: str = Field(
        description="The answer to the user's question"
    )
    source_page: str = Field(
        description="The page number where the answer was found"
    )
    confidence: Literal["high", "medium", "low", "not_found"] = Field(
        description="Confidence level of the answer"
    )
    retrieved_context: str = Field(
        description="The retrieved context used to generate the answer"
    )

print("Output schema defined!")
````

## Step 3: Create the Complete RAG Assistant Class

````
class AIChatAssistant:
    """
    Complete AI PDF Chat Assistant.
    
    This class encapsulates the entire RAG pipeline:
    - PDF loading
    - Text splitting
    - Embedding creation
    - Vector storage
    - Question answering
    - Structured output
    """
    
    def __init__(self, chunk_size=500, chunk_overlap=100):
        """Initialize the assistant with default parameters."""
        self.chunk_size = chunk_size
        self.chunk_overlap = chunk_overlap
        self.chunks = None
        self.vectorstore = None
        self.retriever = None
        self.llm = None
        self.chain = None
        self.simple_chain = None
        self.current_pdf_path = None
        
    def load_pdf(self, pdf_path):
        """Load a PDF and build the complete RAG pipeline."""
        self.current_pdf_path = pdf_path
        
        # Step 1: Load the PDF
        print(f"[1/6] Loading PDF: {pdf_path}")
        loader = PyPDFLoader(pdf_path)
        pages = loader.load()
        
        # Step 2: Split into chunks
        print(f"[2/6] Splitting into chunks...")
        text_splitter = RecursiveCharacterTextSplitter(
            chunk_size=self.chunk_size,
            chunk_overlap=self.chunk_overlap
        )
        self.chunks = text_splitter.split_documents(pages)
        
        # Step 3: Create embeddings model
        print(f"[3/6] Initializing embeddings model...")
        embeddings = HuggingFaceEmbeddings(
            model_name="sentence-transformers/all-MiniLM-L6-v2",
            model_kwargs={"device": "cpu"},
            encode_kwargs={"normalize_embeddings": True}
        )
        
        # Step 4: Create vector store
        print(f"[4/6] Creating vector store...")
        db_path = "../chroma_db"
        if os.path.exists(db_path):
            shutil.rmtree(db_path)
        
        self.vectorstore = Chroma.from_documents(
            documents=self.chunks,
            embedding=embeddings,
            persist_directory=db_path
        )
        
        # Step 5: Create retriever
        self.retriever = self.vectorstore.as_retriever(
            search_kwargs={"k": 4}
        )
        
        # Step 6: Create LLM
        print(f"[5/6] Initializing Mistral via Ollama...")
        self.llm = OllamaLLM(
            model="mistral",
            temperature=0.1,
            num_ctx=4096
        )
        
        # Step 7: Create chains
        print(f"[6/6] Building RAG chains...")
        
        # Chain with structured output
        format_instructions = self._get_format_instructions()
        rag_prompt = PromptTemplate.from_template(
            "You are a precise document analyst. Answer questions using ONLY the provided context.\n\n"
            "## Document Context:\n{context}\n\n"
            "## User Question:\n{question}\n\n"
            "## Strict Rules:\n"
            "1. Answer using ONLY the provided context.\n"
            "2. If the answer is not in the context, set confidence to 'not_found'.\n"
            "3. Do not hallucinate or use outside knowledge.\n"
            "4. Be concise and accurate.\n\n"
            "{format_instructions}"
        ).partial(format_instructions=format_instructions)
        
        self.chain = (
            {"context": self.retriever | self._format_docs, "question": RunnablePassthrough()}
            | rag_prompt
            | self.llm
            | self.parser
        )
        
        # Fallback chain without structured output
        simple_prompt = PromptTemplate.from_template(
            "You are a document analyst. Answer using ONLY the context below.\n\n"
            "## Context:\n{context}\n\n"
            "## Question:\n{question}\n\n"
            "## Instructions:\n"
            "- Use ONLY the provided context.\n"
            "- If the answer is not in the context, say you cannot find it.\n\n"
            "## Answer:\n"
        )
        
        self.simple_chain = (
            {"context": self.retriever | self._format_docs, "question": RunnablePassthrough()}
            | simple_prompt
            | self.llm
        )
        
        return (
            f"PDF loaded successfully!\n"
            f"- Pages: {len(pages)}\n"
            f"- Chunks: {len(self.chunks)}\n"
            f"- Vector store: Ready"
        )
    
    def _get_format_instructions(self):
        """Get format instructions for the output parser."""
        self.parser = PydanticOutputParser(pydantic_object=RAGResponse)
        return self.parser.get_format_instructions()
    
    def _format_docs(self, docs):
        """Format retrieved documents into a context string."""
        return "\n\n".join([
            f"[{i+1}] (Page {doc.metadata.get('page', '?')})\n{doc.page_content}"
            for i, doc in enumerate(docs)
        ])
    
    def ask(self, question):
        """Ask a question and return the answer."""
        if not self.chain:
            return "Please upload a PDF first."
        
        if not question.strip():
            return "Please enter a question."
        
        # Try structured output first
        try:
            result = self.chain.invoke(question)
            return self._format_response(result)
        except Exception as e:
            print(f"Structured output failed: {e}")
        
        # Fallback to simple chain
        try:
            answer = self.simple_chain.invoke(question)
            context = self._format_docs(self.retriever.invoke(question))
            return (
                f"**Answer:** {answer}\n\n"
                f"**Source Page:** See retrieved context\n\n"
                f"**Confidence:** unknown\n\n"
                f"**Retrieved Context:**\n{context}"
            )
        except Exception as e:
            return f"Error: {e}"
    
    def _format_response(self, result):
        """Format the RAGResponse into a display string."""
        confidence_emoji = {
            "high": "[HIGH]",
            "medium": "[MEDIUM]",
            "low": "[LOW]",
            "not_found": "[NOT FOUND]"
        }
        
        return (
            f"**Answer:** {result.answer}\n\n"
            f"**Source Page:** {result.source_page}\n\n"
            f"**Confidence:** {confidence_emoji.get(result.confidence, result.confidence)}\n\n"
            f"**Retrieved Context:**\n{result.retrieved_context}"
        )

print("AIChatAssistant class created!")
````

## Step 4: Build the Gradio Interface

````
# Create the Gradio interface
assistant = AIChatAssistant()

def process_pdf_and_question(pdf_file, question):
    """
    Process the PDF upload and answer the question.
    This function is called when the user clicks 'Ask'.
    """
    if pdf_file is None:
        return "Please upload a PDF file first."
    
    if not question.strip():
        return "Please enter a question."
    
    # Load PDF if it's new or changed
    if assistant.current_pdf_path != pdf_file:
        load_msg = assistant.load_pdf(pdf_file)
        print(load_msg)
    
    # Ask the question
    response = assistant.ask(question)
    return response

# Build the Gradio interface
with gr.Blocks(
    title="AI PDF Chat Assistant",
    theme=gr.themes.Soft()
) as demo:
    gr.Markdown("# AI PDF Chat Assistant")
    gr.Markdown("Upload a PDF and ask questions about its content. The AI will answer based ONLY on the document.")
    
    with gr.Row():
        with gr.Column(scale=1):
            # PDF upload
            pdf_file = gr.File(
                label="Upload PDF",
                file_types=[".pdf"]
            )
            
            # Question input
            question = gr.Textbox(
                label="Your Question",
                placeholder="Ask anything about the PDF...",
                lines=2
            )
            
            # Ask button
            ask_btn = gr.Button("Ask", variant="primary", size="lg")
        
        with gr.Column(scale=2):
            # Output
            output = gr.Textbox(
                label="AI Response",
                lines=15,
                interactive=False
            )
    
    # Connect the button
    ask_btn.click(
        fn=process_pdf_and_question,
        inputs=[pdf_file, question],
        outputs=[output]
    )

print("Gradio interface created!")
````

## Step 5: Launch the Application

````
# Launch the Gradio interface
demo.launch(
    server_name="0.0.0.0",
    server_port=7860,
    share=False
)

print("Application launched!")
print("Open http://localhost:7860 in your browser")
````

## The Complete RAG Pipeline - Summary

Let's review the entire pipeline one more time:

### Phase 1: Indexing (When PDF is uploaded)

```
PDF File
  -> PyPDFLoader (extract text)
  -> RecursiveCharacterTextSplitter (split into chunks)
  -> HuggingFace Embeddings (convert to vectors)
  -> ChromaDB (store vectors)
```

### Phase 2: Querying (When user asks a question)

```
User Question
  -> Embed question (same model)
  -> ChromaDB similarity search (find relevant chunks)
  -> Format chunks as context
  -> Build prompt (context + question + instructions)
  -> Ollama + Mistral (generate answer)
  -> Pydantic Output Parser (structure output)
  -> Display in Gradio GUI
```

### Key Concepts

| Concept | Purpose |
|---------|----------|
| Embeddings | Convert text to numbers for similarity search |
| ChromaDB | Store and search embeddings efficiently |
| Retrieval | Find relevant document chunks |
| Augmentation | Add retrieved context to the prompt |
| Generation | LLM generates answer from augmented prompt |
| Output Parser | Structure the LLM's response |
| Gradio | User interface for the application |

## Troubleshooting

### Common Issues:

1. **"Cannot connect to Ollama"**
   - Make sure Ollama is running: `ollama serve`
   - Check: `curl http://localhost:11434/api/tags`

2. **"Model not found"**
   - Pull the model: `ollama pull mistral`

3. **"PDF not loading"**
   - Check the file path
   - Ensure the PDF is not password-protected

4. **"Parsing failed"**
   - The fallback chain will handle this automatically
   - Try lowering temperature to 0.0

5. **"Slow responses"**
   - Mistral via Ollama runs on CPU by default
   - Use a GPU if available for faster responses
