# Notebook 02: Load PDF

In this notebook, we will learn how to **load a PDF file** and extract its text content.

## What You Will Learn

- How to use LangChain's PyPDFLoader
- How to read the entire content of a PDF
- How to access metadata (like page numbers)
- Why loading PDFs is the first step in a RAG pipeline

## What is RAG?

**RAG** stands for **Retrieval-Augmented Generation**. It's a technique that:

1. **Retrieves** relevant information from a document
2. **Augments** a prompt with that information
3. **Generates** an answer using a language model

The first step in RAG is always: **load the document**.

## Why Use LangChain?

**LangChain** is a framework that provides tools to work with LLMs. It has built-in document loaders that can handle PDFs, Word documents, web pages, and more.

## Step 1: Import Required Libraries

```
from langchain_community.document_loaders import PyPDFLoader

# PyPDFLoader is a LangChain tool that reads PDF files
# It extracts text from each page and stores it as a 'Document' object
```

## Step 2: Load the PDF

We use `PyPDFLoader` to load our sample PDF. Replace `'sample.pdf'` with the path to your PDF.

```
# Create a PDF loader
# Replace '../data/sample.pdf' with the path to your own PDF
loader = PyPDFLoader("../data/sample.pdf")

# Load the PDF - this reads all pages and returns a list of Document objects
pages = loader.load()

# Print how many pages were loaded
print(f"Total pages loaded: {len(pages)}")
print(f"Type of each page: {type(pages[0])}")
```

## Step 3: Understand the Document Object

Each page is a **Document** object with two properties:

- `page_content`: The text extracted from the page
- `metadata`: Information about the page (source file, page number, etc.)

```
# Let's look at the first page
first_page = pages[0]

print("=" * 60)
print("PAGE CONTENT (first 300 characters):")
print("=" * 60)
print(first_page.page_content[:300])

print("\n" + "=" * 60)
print("METADATA:")
print("=" * 60)
print(first_page.metadata)
```

## Step 4: Look at All Pages

Let's see a preview of all pages and their metadata.

```
# Display a summary of each page
for i, page in enumerate(pages):
    print(f"\n{'=' * 60}")
    print(f"Page {i + 1}")
    print(f"{'=' * 60}")
    print(f"Characters: {len(page.page_content)}")
    print(f"Preview: {page.page_content[:100]}...")
    print(f"Metadata: {page.metadata}")
```

## Step 5: Handle PDFs That Cannot Be Loaded

Sometimes PDFs might fail to load due to:

- Corrupted files
- Password-protected PDFs
- Scanned images (no text layer)

Here's how to handle errors gracefully.

```
try:
    loader = PyPDFLoader("../data/nonexistent.pdf")
    pages = loader.load()
    print(f"Loaded {len(pages)} pages.")
except Exception as e:
    print(f"Error loading PDF: {e}")
    print("This is expected - the file doesn't exist.")
    print("In a real application, you would show this error to the user.")
```

## Key Takeaways

1. **PyPDFLoader** reads PDF files and converts each page into a `Document` object
2. Each `Document` has `page_content` (the text) and `metadata` (page info)
3. The metadata includes the source file path and page number
4. We need to **split** these documents into smaller chunks next (Notebook 03)

## Why Do We Need to Split?

LLMs have a **context window** limit (e.g., 4096 tokens for Mistral). If a PDF page is too large, we need to split it into smaller chunks that fit within this limit.
