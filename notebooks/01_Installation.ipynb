## Step 1: Install Ollama

**Ollama** is a tool that lets you run Large Language Models (LLMs) locally on your computer.

### On Linux:
```bash
curl -fsSL https://ollama.com/install.sh | sh
```

### On macOS:
```bash
brew install ollama
```

### On Windows:
Download from: https://ollama.com/download

## Step 2: Start Ollama and Pull Mistral

After installing Ollama, start the Ollama server and download the **Mistral** model.

```bash
# Start Ollama server (in a terminal)
ollama serve

# In another terminal, pull the Mistral model
ollama pull mistral
```

The Mistral model is approximately 4.1 GB. Make sure you have enough disk space.

## Step 3: Create a Virtual Environment

It's best practice to use a virtual environment to keep your project dependencies isolated.

```bash
# Create a virtual environment
python3 -m venv venv

# Activate it (Linux/macOS)
source venv/bin/activate

# Activate it (Windows)
# venv\Scripts\activate
```

## Step 6: Verify Ollama Connection

Let's check if Ollama is running and the Mistral model is available.

````python
import requests

try:
    response = requests.get("http://localhost:11434/api/tags", timeout=5)
    if response.status_code == 200:
        models = response.json().get("models", [])
        print(f"Ollama is running! Available models:")
        for model in models:
            print(f"  - {model.get('name', 'unknown')}")
    else:
        print("Ollama returned an unexpected status code.")
except requests.exceptions.ConnectionError:
    print("ERROR: Cannot connect to Ollama!")
    print("Make sure you have started Ollama with: ollama serve")
    print("And pulled the Mistral model with: ollama pull mistral")
````
